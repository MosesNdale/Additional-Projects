```python
#From module and importing module

from pygame import mixer
from tkinter import Tk
from tkinter import Label
from tkinter import Button
from tkinter import filedialog
from selenium import webdriver
import tkinter.messagebox
from tkinter import *
from datetime import datetime
from mutagen.mp3 import MP3
from mutagen.wave import WAVE





#Sets the current volume to medium at its default position
current_volume= float(0.5)


#Subroutines
def play_song():
    filename=filedialog.askopenfilename(initialdir="C:/",title="Please select a file")
    current_song=filename
    song_title=filename.split("/")
    song_title=filename[-1]
    current_position_of_song()
    song_length(filename)
    date_time()
   
    try:
        mixer.init()
        mixer.music.load(current_song)
        mixer.music.set_volume(current_volume)
        mixer.music.play()
        functions_title_label.config(fg="black",text="Loading...")
        song_title_label.config(fg="green",text="Song: "+str(current_song),pady=4,borderwidth=3, relief="ridge")
        volume_label.config(fg="green", text="Volume: "+str(current_volume),pady=4)
    except Exception as e:
        print(e)
        song_title_label.config(fg="red", text="Error playing this song")

def reduce_volume():
    try:
        global current_volume
        if current_volume <=0:
            volume_label.config(fg="red",text="The volume is muted")
            return
        current_volume=current_volume-float(0.1)
        current_volume=round(current_volume,1)
        mixer.music.set_volume(current_volume)
        volume_label.config(fg="green",text="Volume :"+str(current_volume))
    except Exception as e:
        print(e)
        song_title_label.configure(fg="red",text="Can't reduce volume without music")
        tkinter.messagebox.showerror("Music Player", "There is no music selected from files to reduce volume")

def increase_volume():
    try:
        global current_volume
        if current_volume >=1:
            volume_label.config(fg="green",text="The volume is at max")
            return
        current_volume=current_volume+float(0.1)
        current_volume=round(current_volume,1)
        mixer.music.set_volume(current_volume)
        volume_label.config(fg="green",text="Volume :"+str(current_volume))
    except Exception as e:
        print(e)
        song_title_label.config(fg="red",text="Can't increase volume without music")
        tkinter.messagebox.showerror("Music Player", "There is no music selected from files to increase volume")

def pause():
    try:
        mixer.music.pause()
        functions_title_label.config(fg="red",text="Music is paused")
    except Exception as e:
        print(e)
        song_title_label.config(fg="red",text="There is no music to pause")
        tkinter.messagebox.showerror("Music Player", "Can't pause music without selection from files")

def unpause():
    try:
        mixer.music.unpause()
        functions_title_label.config(fg="green",text="Music is resumed")
    except Exception as e:
        print(e)
        song_title_label.config(fg="red",text="There is no music to resume")
        tkinter.messagebox.showerror("Music Player", "There is no music selected from files to resume")

def play():
    try:
        mixer.music.play()
        functions_title_label.config(fg="orange",text="Music is playing")
    except Exception as e:
        print(e)
        song_title_label.config(fg="red",text="There is no music to play")
        tkinter.messagebox.showerror("Music Player", "Can't play music without selection from files or use the search engine below")

def stop():
    try:
        mixer.music.stop()
        functions_title_label.config(fg="red",text="Music has stopped")
    except Exception as e:
        print(e)
        song_title_label.config(fg="red",text="There is no music to stop")
        tkinter.messagebox.showerror("Music Player", "There is no music selected from files to stop")

def about_us():
    tkinter.messagebox.showinfo("About Mr Music (MP3)", "This is a music player built by using Python Tkinter to listen to music by clicking on the files or browse on the search engine by @Moses Ndale")        
   
def date_time():
    now= datetime.now()
    date=now.strftime("%d/%m/%Y")
    date_label.config(text="Date: "+str(date))
    time= now.strftime("%I:%M %p")
    time_label.config(text="Time: "+str(time))
    time_label.after(200,date_time)

def convert(seconds):
    hours= seconds // 3600
    seconds %= 3600
    mins= seconds // 60
    seconds %= 60
    return hours, mins, seconds    

def current_position_of_song():
    mixer.init()
    current_pos= mixer.music.get_pos()
    current_pos= int(current_pos * 0.001)
    h,m,s = convert(current_pos)
    position_label.config(text="Duration- "+str(h)+"h"+":"+str(m)+"m"+":"+str(s)+"s",pady=3,borderwidth=3, relief="ridge")
    position_label.after(1000,current_position_of_song)
   
def song_length(filename):
    audio = MP3(filename)
    audio_info = audio.info
    length_in_secs = int(audio_info.length)
    hours, mins, seconds = convert(length_in_secs)
    length_label.config(text="Length- "+str(hours)+"h"+":"+str(mins)+"m"+":"+str(seconds)+"s",pady=3,borderwidth=3, relief="ridge")
   
       

def search_result():
    bot=webdriver.Chrome()
    bot.maximize_window()
    url="https://www.google.co.uk/"
    bot.get(url)

def soundcloud():
    bot=webdriver.Chrome()
    bot.maximize_window()
    url="https://soundcloud.com/"
    bot.get(url)


def fullscreen():
    master.state("zoomed")    
                     



#Main Screen
master = Tk()
master.title("Music Player")
photo= PhotoImage(file="C:\\Users\\mndal\\OneDrive\\Pictures\\410a52451ba7a53ba5b0d7f47d8bf3db.png")
background_label= Label(master, image=photo)
background_label.place(x=0, y=0, relwidth=1, relheight=1)



#Labels
Label(master,text="Mr Music",font=("Fixedsys",15),fg="white",bg="grey",borderwidth=1,relief="solid").grid(sticky="N",row=0,padx=120)
Label(master,text="Welcome, select a music track you want to play",font=("Fixedsys",12),fg="white",bg="grey",borderwidth=1,relief="solid").grid(sticky="N",row=1)
Label(master,text="Volume",font=("Fixedsys",12),fg="white",bg="grey",borderwidth=1,relief="solid").grid(sticky="N",row=4)
song_title_label= Label(master,font=("Fixedsys",12),borderwidth=1,relief="solid")
song_title_label.grid(stick="N",row=3,padx=180)
functions_title_label= Label(master,font=("Fixedsys",12),borderwidth=1,relief="solid",pady=3)
functions_title_label.grid(stick="E",row=2)
volume_label= Label(master,font=("Fixedsys",12),borderwidth=1,relief="solid")
volume_label.grid(sticky="N", row=5)
Label(master,text="  Search Engine:  ",font=("Fixedsys",16),fg="#3a34eb",bg="white",borderwidth=1,relief="solid").grid(row=8,sticky="W")
date_label= Label(master,font=("Fixedsys",12),borderwidth=1,relief="solid")
date_label.grid(sticky="W",row=0)
time_label= Label(master,font=("Fixedsys",12),borderwidth=1,relief="solid")
time_label.grid(sticky="E",row=0)
position_label= Label(master,font=("Fixedsys",12),borderwidth=1,relief="solid")
position_label.place(x=200, y=136)
length_label= Label(master,font=("Fixedsys",12),borderwidth=1,relief="solid")
length_label.place(x=503, y=136)

#Buttons
Button(master, text="  Files  ", font=("Fixedsys",12),fg="gold",bg="grey",activebackground="gold",activeforeground="grey",command=play_song,borderwidth=1,relief="solid").grid(row=2, sticky="W")
Button(master, text="Google",font=("Fixedsys",12),fg="turquoise",bg="white",activeforeground="white",activebackground="turquoise",command=search_result,borderwidth=1,relief="solid").grid(row=8,sticky="N")
Button(master, text="Listen Soundcloud", font=("Fixedsys",12),fg="#a34f14",bg="white",activeforeground="white",activebackground="#a34f14",command=soundcloud,borderwidth=1,relief="solid").grid(row=8,sticky="E")
Button(master, text="Pause", font=("Fixedsys",12),fg="#801127",bg="grey",activebackground="white",activeforeground="#801127",command=pause,borderwidth=1,relief="solid").grid(row=3,sticky="E")
Button(master, text="Play", font=("Fixedsys",12),fg="#0f8c50",bg="grey",activebackground="white",activeforeground="#0f8c50",command=play,borderwidth=1,relief="solid").grid(row=3,sticky="W")
Button(master, text="Stop", font=("Fixedsys",12),fg="white",bg="#801127",activebackground="white",activeforeground="#801127",command=stop,borderwidth=1,relief="solid").grid(row=7,sticky="W")
Button(master, text="Resume", font=("Fixedsys",12),fg="white",bg="#0f8c50",activebackground="white",activeforeground="#0f8c50",command=unpause,borderwidth=1,relief="solid").grid(row=7,sticky="E")
Button(master, text="-", font=("Fixedsys",12),fg="#801127",bg="grey",activeforeground="white",activebackground="#801127",command=reduce_volume,width=5,borderwidth=1,relief="solid").grid(row=5,sticky="W")
Button(master, text="+", font=("Fixedsys",12),fg="#0f8c50",bg="grey",activeforeground="white",activebackground="#0f8c50",command=increase_volume,width=5,borderwidth=1,relief="solid").grid(row=5,sticky="E")
Button(master, text="Info", font=("Fixedsys",12),fg="black",bg="white",activeforeground="white",activebackground="black",command=about_us,borderwidth=1,relief="solid").grid(row=2,sticky="N")
Button(master, text="Maximize", font=("Fixedsys",12),fg="black",bg="white",activeforeground="white",activebackground="#0f8c50",command=fullscreen,borderwidth=1,relief="solid").grid(row=9,sticky="N")

#Runs the program event
master.mainloop()
```
